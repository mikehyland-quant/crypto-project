```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from Methods_xlWings import *
from Monte_Carlo import *
```

```python
    # ----------------------
    # Function to simulate GBM with stochastic volatility
    # ----------------------
class Black_Scholes_MC(Monte_Carlo):
    def __init__(self):
        pass

    def simulate_mc(
                    self, 
                    S,     # initial stock price
                    Z,
                    daily_drift,
                    daily_vol,
                    steps
                ):
        
        for t in range(1, steps):
            S[t] = S[t-1] * np.exp(daily_drift - 0.5 * daily_vol + np.sqrt(daily_vol) * Z[t])

        return S
```

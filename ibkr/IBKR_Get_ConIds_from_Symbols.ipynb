```python
from ib_insync import *

symbols = [
    'FENY', 'FHLC', 'FIDU', 'FREL', 'FSTA', 'FTEC', 'FUTY',
    'IDU', 'IYE', 'IYF', 'IYH', 'IYR', 'IYW',
    'SCHH', 'USRT',
    'VDC', 'VDE', 'VFH', 'VGT', 'VHT', 'VNQ', 'VPU',
    'XLE', 'XLF', 'XLI', 'XLK', 'XLP', 'XLRE', 'XLU', 'XLV'
]

ib = IB()

await ib.connectAsync('127.0.0.1', 7496, clientId=2)

for sym in symbols:
    contract = Stock(sym, 'SMART', 'USD')

    qualified = await ib.qualifyContractsAsync(contract)

    if qualified:
        c = qualified[0]
        print(f"{sym:<5} {c.conId}")
    else:
        print(f"{sym:<5} NOT FOUND")

ib.disconnect()
```

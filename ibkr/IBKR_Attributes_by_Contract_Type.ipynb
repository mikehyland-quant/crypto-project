```python
from datetime import datetime

from ib_insync import *
from Class_IBKR_IB import *
ibkr = IBKR_IB()

ibkr.ib.disconnect()

clientId = int(datetime.now().strftime("%H%M%S"))
await ibkr.ib.connectAsync('127.0.0.1', 7496, clientId=clientId)

print("Connected:", ibkr.ib.isConnected())
```

```python
async def display_output(contract):
    dict_ = vars(contract)
    for k, v in dict_.items():
        print(type(v), k, v)

    print('\n', '\n')


    details_list = await ibkr.ib.reqContractDetailsAsync(contract)
    for details in details_list:
        dict_ = vars(details)

        for k, v in dict_.items():
            print(type(v), k, v)

    print('\n', '\n')


    ticker = ibkr.ib.reqMktData(contract, '', False, False)
    await asyncio.sleep(2)   # allow data to arrive

    dict_ = vars(ticker)
    for k, v in dict_.items():
        print(type(v), k, v)
```

```python
# BTC Spot

contract = await ibkr.contract_by_conId(541686651)
# contract_by_conId includes qualifyContractsAsync
# no need for await ibkr.ib.qualifyContractsAsync(contract) 

await display_output(contract)
```

```python
#  Stock

contract = Stock('IBIT', 'SMART', 'USD')
await ibkr.ib.qualifyContractsAsync(contract)

await display_output(contract)
```

```python
# Stock Option

contract = Option(symbol='IBIT',
                  lastTradeDateOrContractMonth='20260618',
                  strike=50.0,
                  right='C',
                  exchange='SMART',
                  currency='USD')
await ibkr.ib.qualifyContractsAsync(contract)

await display_output(contract)
```

```python
#  Future

contract = Future(symbol='BRR', lastTradeDateOrContractMonth='202606', exchange='CME', currency='USD')
await ibkr.ib.qualifyContractsAsync(contract)

await display_output(contract)
```

```python
# Futures Option

contract = FuturesOption(
    symbol='BRR',
    lastTradeDateOrContractMonth='20260626',
    strike=65000.0,
    right='C',
    exchange='CME',
    currency='USD'
)
await ibkr.ib.qualifyContractsAsync(contract)

await display_output(contract)
```

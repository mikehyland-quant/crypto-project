```python
# --- system setup ---
import sys
import os
sys.path.append(os.path.abspath(".."))
```

```python
from ib_insync import Option, FuturesOption
from ibkr.Class_IBKR_IB import IBKR_IB
ibkr = IBKR_IB(port=7496)

async def start_ibkr():
    await ibkr.connect()
    print("IBKR connected:", ibkr.ib.isConnected())

await start_ibkr()
```

```python
underlying_conId = 751356962 
```

```python
underlying_contract = await ibkr.contract_by_conId(underlying_conId)
```

```python
underlying_details = await ibkr.ib.reqContractDetailsAsync(underlying_contract)
```

```python
print(underlying_details)
```

```python
option_chain = await ibkr.ib.reqSecDefOptParamsAsync(underlyingSymbol=underlying_details[0].contract.symbol ,
                                                       futFopExchange=underlying_details[0].contract.exchange,
                                                       underlyingSecType=underlying_details[0].contract.secType,
                                                       underlyingConId=underlying_conId)
```

```python
print(option_chain)
```

```python
print(option_chain[0].expirations)
```

```python
print(option_chain[0].strikes)
```

```python
option_contracts = []

for expiry in option_chain[0].expirations:
    for strike in option_chain[0].strikes:
        for right in ["C", "P"]:
            opt = FuturesOption(
                lastTradeDateOrContractMonth=expiry,
                strike=float(strike),
                right=right,
                symbol=underlying_details[0].contract.symbol,                    # from underlying future details
                exchange=underlying_details[0].contract.exchange,
                currency=underlying_details[0].contract.currency
            )
            option_contracts.append(opt)
```

```python
qualified_options = await ibkr.ib.qualifyContractsAsync(*option_contracts)

# might take 15 seconds or more
```

```python
print(qualified_options)
```

```python
rows = []

for opt in qualified_options:
    rows.append({
        "expiry": opt.lastTradeDateOrContractMonth,
        "strike": opt.strike,
        "right": opt.right,
        "option_conId": opt.conId,
        #"symbol": opt.symbol,
        #"localSymbol": opt.localSymbol,
        #"exchange": opt.exchange,
        #"currency": opt.currency,
        #"tradingClass": opt.tradingClass,
        #"multiplier": opt.multiplier
    })
```

```python
import pandas as pd
df_options = pd.DataFrame(rows)

df_options.insert(0, "underlying_conId", underlying_conId)

df_options
```

```python
call_df = df_options[df_options["right"] == "C"]
put_df = df_options[df_options["right"] == "P"]

options_df = pd.merge(call_df, put_df, on=['underlying_conId', "expiry", "strike"], suffixes=("_call", "_put"))
options_df.drop(columns=["right_call", "right_put"], inplace=True)

options_df
```

```python
options_df.to_csv("options_conIds.csv", index=False)
```

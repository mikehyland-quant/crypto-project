```python
# ============================================================
# INPUT SYMBOLS AS A LIST WITH ANCHOR SYMBOL LAST
# ============================================================

sorted_symbols_list = ['SCHH', 'USRT']

moving_avg_days = 20
```

```python
# --- system setup ---
import sys
import os
from pathlib import Path 
sys.path.append(os.path.abspath(".."))

# from datetime import date
import asyncio
import numpy as np
import pandas as pd
```

```python
filename = "_".join(sorted_symbols_list)
filename = filename + ".csv"
file_path = Path("enhanced_prices/") / filename

df = pd.read_csv(file_path, index_col=0)
```

```python
anchor = sorted_symbols_list[-1]
for sym in sorted_symbols_list:
    existing_col_name = f"to {anchor} / to {sym}"
    new_col_name = f"{anchor} / {sym} moving avg"
    yday_col_name = "yday " + new_col_name
    df[new_col_name] = df[existing_col_name].rolling(moving_avg_days).mean()
    df[yday_col_name] = df[new_col_name].shift(1)
    df[f"to {sym} unit price"] = df[yday_col_name] * df[f"to {sym} price"]
```

```python
for sym in sorted_symbols_list:
    df[f"{sym} unit price pct diff"] = np.log(df[f"to {sym} unit price"] / df[f"to {anchor} unit price"])
```

```python
non_anchor = sorted_symbols_list[0]
df['unit price pct diff'] = df[f"{non_anchor} unit price pct diff"] - df[f"{anchor} unit price pct diff"]
```

```python
filename = filename.replace(".csv", f"_{moving_avg_days}.csv")

file_path = Path("unit_prices/") / filename

df.to_csv(file_path)

print()
print('finished')
print()
```

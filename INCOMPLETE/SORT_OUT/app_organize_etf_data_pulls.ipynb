```python
time_name = first_name.split("_")

t = datetime.strptime(time_name[3], "%H-%M-%S")   # ⬅️ NOT .time()
t_plus_10 = t + timedelta(seconds=600)

time_str = t_plus_10.strftime("%H-%M-%S")
file_name = time_name[0] + "_" + time_name[1] + "_" + time_name[2] + "_" + time_str
```
